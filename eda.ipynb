```python
import matplotlib.pyplot as plt
import pandas as pd
data_counts = df["Category"].value_counts()
plt.figure(figsize=(16, 16))
data_counts.plot(
    kind="pie",
    autopct="%1.1f%%",  # Automatically calculates and formats percentages
    startangle=140,  # Rotates the pie for better aesthetics
    colors=["#5DADE2", "#48C9B0", "#F4D03F", "#E74C3C"],  # Optional custom colors
)
# 3. Clean up the presentation
plt.title("Distribution of Category", fontsize=14, pad=15)
plt.show()
```
